```python


ImageFile.LOAD_TRUNCATED_IMAGES = True

img_width, img_height = 150, 150
batch_size = 32

train_datagen = ImageDataGenerator()

train_generator = train_datagen.flow_from_directory( "/content/data/Bone_Fracture_Binary_Classification/Bone_Fracture_Binary_Classification/train", target_size=(img_width, img_height), batch_size=batch_size,class_mode='binary',shuffle=False)

validation_generator = train_datagen.flow_from_directory('/content/data/Bone_Fracture_Binary_Classification/Bone_Fracture_Binary_Classification/val',target_size=(img_width, img_height),batch_size=batch_size,class_mode='binary')

test_generator = train_datagen.flow_from_directory( '/content/data/Bone_Fracture_Binary_Classification/Bone_Fracture_Binary_Classification/test',target_size=(img_width, img_height),batch_size=batch_size, class_mode='binary',shuffle=False)
```

```python
model = keras.Sequential()

model.add(keras.layers.Conv2D(32, (3, 3),activation='relu',input_shape=(img_width, img_height, 3)))

model.add(keras.layers.MaxPooling2D((2, 2)))

model.add(keras.layers.Conv2D(64, (3, 3),activation='relu'))

model.add(keras.layers.MaxPooling2D((2, 2)))

model.add(keras.layers.Conv2D(64, (3, 3),activation='relu'))

model.add(keras.layers.MaxPooling2D((2, 2)))

model.add(keras.layers.Flatten())

dropout = keras.layers.Dropout(0.5)
model.add(keras.layers.Dense(128, activation='relu'))
model.add(keras.layers.Dense(1, activation='sigmoid'))
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
callback=keras.callbacks.EarlyStopping(monitor='val_loss',patience=3)
model.fit(train_generator,epochs=50,validation_data=validation_generator,callbacks=[callback])
```

```python
model.evaluate(test_generator)
```

```python
test_generator.reset()

train_generator.reset()
y_pred_val= model.predict(test_generator)
y_pred_train= model.predict(train_generator)
y_train_true=train_generator.classes
y_pred_train = (y_pred_train > 0.5).astype(int).ravel()
y_pred = (y_pred_val > 0.5).astype(int).ravel()


y_true = test_generator.classes
print("Classification_report for train")
print(classification_report( y_train_true,y_pred_train))
print("classification report for test")
print(classification_report( y_true,y_pred))
print("Confusion Matrix for train:")
cm1 = confusion_matrix(y_train_true, y_pred_train)
print(cm1)
print("Confusion Matrix for test:")
cm = confusion_matrix(y_true, y_pred)


print(cm)
```
